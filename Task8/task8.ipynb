```python
from matplotlib import pyplot as plt
import pandas as pd 
```

```python
plt.pie(data['Customer Type'].value_counts(),labels=data['Customer Type'].unique(),wedgeprops={'edgecolor':'black'},autopct='%1.1f%%')
plt.title('Percentages of Customer Types')
plt.show()
```

```python
satisfaction = data['satisfaction'].value_counts()
plt.pie(satisfaction.values,labels=(satisfaction.index),autopct='%1.1f%%',wedgeprops={'edgecolor':'black'})
plt.title('Overall Satisfaction Level ')
```

```python
classes = data['Class'].value_counts()
plt.bar(classes.index,classes.values)
plt.ylabel('Number of Passengers')
plt.xlabel('Class')
plt.title('Number of Passengers per Class')
```

```python
classes_satisfaction = pd.crosstab(data['Class'], data['satisfaction'])
classes_satisfaction.plot(kind='bar')
plt.xlabel('Class')
plt.ylabel('Number of Passengers')
plt.grid(True)
plt.title('Satisfaction by Class')
plt.legend()
```

```python
bins=[10,20,30,40,50,60,70,80,90]
plt.hist(data['Age'],edgecolor='black',bins=bins)
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title('Passengers by Age')
```

```python
satisfied_age = data[ data['satisfaction'] == 'satisfied' ]['Age']
dissatisfied_age = data[ data['satisfaction'] == 'neutral or dissatisfied' ]['Age']
satisfied_age.plot(kind='hist',linewidth=1,edgecolor='black',label='satisfied')
dissatisfied_age.plot(kind='hist',linewidth=1,edgecolor='black',alpha=.8,label='neutral or unsatisfied')
plt.title("Satisfaction by Age")
plt.xlabel("Age")
plt.legend()
```

```python
services = data[['Inflight wifi service',
       'Departure/Arrival time convenient', 'Ease of Online booking',
       'Gate location', 'Food and drink', 'Online boarding', 'Seat comfort',
       'Inflight entertainment', 'On-board service', 'Leg room service',
       'Baggage handling', 'Checkin service', 'Inflight service',
       'Cleanliness']]
plt.title("Average Services Rating")
serv_list = services.columns.tolist()
plt.grid(True)
plt.barh(serv_list,services.mean())
```

```python
plt.scatter(data['Departure Delay in Minutes'],data['Flight Distance'])
plt.title('Flight Distance and Departure Delay')
plt.xlabel('Departure Delay in Minutes')
plt.ylabel('Flight Distance')
```

```python
plt.scatter(data['Departure Delay in Minutes'], data['Arrival Delay in Minutes'])
plt.xlabel('Departure Delay in Minutes')
plt.ylabel('Arrival Delay in Minutes')
plt.grid(True)
plt.title("Relationship between Departure Delay in Minutes and Arrival Delay")
```

```python
serv_satis = data.groupby('satisfaction')[serv_list].mean()
serv_satis.plot(kind='barh',edgecolor='black',linewidth=.5,width=.8)
plt.xlabel("Service Average Rating")
plt.title("Average Service Rating by Satisfaction Level")
plt.legend(loc="center left", bbox_to_anchor=(1.02, 0.5))
```

```python
l = data.groupby('Class')[serv_list].mean()
l.plot(kind = "barh",width= .91,edgecolor = 'black', linewidth=.5)
plt.legend(loc="center left", bbox_to_anchor=(1.02, 0.5))
plt.xlabel("Service Average Rating")
plt.title("Average Service Rating by Class")
```
